```python
import pandas as pd
# df可以理解为python里的字典嵌套
df = pd.read_csv("./data/1960-2019全球GDP数据.csv", encoding="GBK")
print(df.dtypes)
df.head()
```

```python
china_gdp = df[df.country == "中国"]
china_gdp.head(10)
```

```python
# 由于year现在是索引，所以绘图时不需要指定x参数，pandas会自动使用索引作为x轴
# 正常情况是:china_gdp.plot(x='year',y='GDP', title='GDP over Time')
china_gdp.plot(y='GDP', title='GDP over Time')
```

```python
# 等价于china_gdp.plot(y='GDP', title='GDP over Time')
china_gdp.GDP.plot()
```

```python
# 使用同样的方法画出日本的GDP变化曲线，和中国的GDP变化曲线进行对比
# 1. DataFrame 的列作为属性访问
# 在 Pandas 中，df.country 这种写法是合法的，因为 Pandas 提供了类似属性的访问方式。
# 在 Pandas 中，如果 DataFrame 的列名是有效的 Python 变量名（即没有空格、特殊字符，且不与 Python 关键字冲突），
# 你可以像访问对象属性一样使用 df.country 来访问 country 这一列。
# 这是 Pandas 的一个特性，称为 属性访问，它允许你通过 df.column_name 来访问 DataFrame 中的列。
# 这种方式是为了提高代码的可读性和便捷性。

# 2. 布尔索引（Boolean Indexing）

# df[df.country == "中国"] 这一部分是 Pandas 中的 布尔索引。它的含义是：

# df.country == "中国" 会生成一个布尔 Series，表示每一行的 country 列是否等于 "中国"。

# 然后，df[df.country == "中国"] 将会用这个布尔 Series 作为索引，返回所有符合条件的行，即 country 列为 "中国" 的行。

# 3. 布尔索引的工作原理

# 布尔索引的工作方式是利用布尔值的 True/False 来选择 DataFrame 中的行。这是 Pandas 中的一种非常强大的索引方式，可以通过逻辑条件筛选出符合条件的行。

# 具体来说，df.country == "中国" 会返回一个布尔值的 Series：

# 0     True
# 1    False
# 2     True
# ...


# 然后，df[df.country == "中国"] 就相当于在 df 中选出所有 True 的行。
jp_gdp = df[df.country == '日本'].set_index('year')
# jp_gdp.plot(y='GDP')
china_gdp.GDP.plot()
```

```python
import pandas as pd
import matplotlib
# matplotlib.use('TkAgg')
import matplotlib.pyplot as plt
df = pd.read_csv("./data/1960-2019全球GDP数据.csv", encoding="GBK")

plt.rcParams['font.sans-serif'] = ['Arial Unicode MS', 'PingFang SC', 'Heiti SC']
plt.rcParams['axes.unicode_minus'] = False

jp_gdp = df[df.country == "日本"].set_index("year")
china_gdp = df[df.country == "中国"].set_index("year")
us_gdp = df[df.country == "美国"].set_index("year")

# 对指定列名进行修改
china_gdp.rename(columns={"GDP":"中国"}, inplace=True)
us_gdp.rename(columns={"GDP":"美国"}, inplace=True)
jp_gdp.rename(columns={"GDP":"日本"}, inplace=True)

# 画图
jp_gdp.日本.plot(legend=True)
china_gdp.中国.plot(legend=True)
us_gdp.美国.plot(legend=True)
```
