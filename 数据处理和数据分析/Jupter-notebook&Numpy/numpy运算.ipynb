```python
import numpy as np

# 数组的算数运算是按照元素的。新的数组被创建并且被结果填充
a = np.array([20,30,40,50])
b = np.arange(4)
c = a - b
print('数组a:',a)
print('数组b:',b)
print('数组运算a-b:',c)
print('-'*34)

# 2 矩阵乘法 矩阵对应元素的乘法（multiplication by element-wise）
d = np.random.randint(1,5,size=(3,4))
print('d:',d)
e = np.random.randint(6,10,size=(3,4))
print('e:',e)
f = d * e
print('f:',f)

g = np.multiply(d,e)
print('g:',g)
print('-'*34)

x = np.array([[1,2,3],[4,5,6]])
y = np.array([[6,23],[-1,7],[8,9]])

print('x.dot(y):',x.dot(y))
print('np.dot(x,y):',np.dot(x,y))


```
